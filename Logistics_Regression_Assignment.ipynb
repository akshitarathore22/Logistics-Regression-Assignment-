Q1. What is Simple Linear Regression (SLR)? Explain its purpose.

A1. Simple Linear Regression (SLR) is a statistical method used to examine and describe the relationship between two quantitative variables by fitting a straight-line equation to observed data. In this technique, one variable is treated as the independent or explanatory variable, while the other is treated as the dependent or response variable. The core idea behind SLR is that changes in the dependent variable can be explained, at least partly, by changes in the independent variable, and this relationship can be approximated using a linear function. The regression line represents the best possible straight-line fit through the data points by minimizing the differences between the actual observed values and the values predicted by the model, which are known as errors or residuals.

The purpose of Simple Linear Regression is not only to identify whether a relationship exists between the two variables, but also to measure the direction and strength of that relationship. By estimating the slope of the regression line, SLR indicates how much the dependent variable is expected to increase or decrease for a one-unit change in the independent variable. This makes it a valuable tool for explaining trends and patterns observed in real-world data. Additionally, Simple Linear Regression is widely used for prediction, allowing researchers, analysts, and decision-makers to estimate the value of an unknown dependent variable based on a given value of the independent variable. Overall, SLR provides a simple yet powerful way to analyze relationships, draw conclusions, and support data-driven decisions in fields such as economics, business, science, and social research.



Q2. What are the key assumptions of Simple Linear Regression?

A2. The key assumptions of Simple Linear Regression are the conditions that must be satisfied for the results of the model to be valid and reliable. First, it is assumed that there is a linear relationship between the independent variable and the dependent variable, meaning that the change in the dependent variable is proportional to the change in the independent variable and can be represented by a straight line. Another important assumption is independence of errors, which means that the observations and their associated error terms are not related to each other; the value of one observation should not influence another. Simple Linear Regression also assumes homoscedasticity, meaning that the variance of the error terms remains constant across all levels of the independent variable, rather than increasing or decreasing as the value of the independent variable changes. In addition, the model assumes that the error terms are normally distributed, especially for the purpose of hypothesis testing and confidence interval estimation. Finally, it is assumed that the independent variable is measured without error or with negligible error. When these assumptions are reasonably satisfied, Simple Linear Regression provides accurate estimates and meaningful inferences.

Q5.  What is the method of least squares in linear regression?

A5. The method of least squares in linear regression is a mathematical approach used to estimate the best-fitting regression line for a given set of data points. The main idea behind this method is to choose the values of the regression coefficients in such a way that the total error between the observed values and the values predicted by the regression line is as small as possible. Specifically, it minimizes the sum of the squares of the residuals, where a residual is the difference between an actual observed value of the dependent variable and the corresponding value predicted by the model. Squaring the residuals ensures that both positive and negative errors contribute positively and gives greater weight to larger errors. By minimizing this sum, the method of least squares produces a unique line that best represents the overall trend in the data and provides reliable estimates for prediction and analysis.


Q6. What is Logistic Regression? How does it differ from Linear Regression?

A6. Logistic regression is a statistical technique used to model the relationship between one or more independent variables and a **binary dependent variable**, meaning the outcome has only two possible categories such as success or failure, yes or no, or pass or fail. Instead of predicting a continuous value, logistic regression estimates the **probability** that a particular outcome will occur. It does this by applying a logistic (sigmoid) function to a linear combination of the independent variables, which ensures that the predicted values lie between 0 and 1. This makes logistic regression especially useful for classification problems where the goal is to assign observations to one of two groups based on the values of the predictors.

Linear regression, on the other hand, is used when the dependent variable is **continuous**, and it models the relationship between variables by fitting a straight line to the data. The predicted values in linear regression can take any real number, including negative values or values greater than one, which makes it unsuitable for classification tasks. Another key difference lies in the method of estimation: linear regression typically uses the method of least squares, while logistic regression uses maximum likelihood estimation to find the best-fitting model. In summary, logistic regression differs from linear regression in terms of the type of dependent variable it handles, the form of the model, and the interpretation of its results, making each suitable for different kinds of real-world problems.


Q7.  Name and briefly describe three common evaluation metrics for regression
models.

A7. Three common evaluation metrics used for regression models are Mean Absolute Error (MAE), Mean Squared Error (MSE), and R-squared. Mean Absolute Error measures the average of the absolute differences between the actual observed values and the values predicted by the model, giving a clear idea of how far the predictions are from the true values on average. Mean Squared Error calculates the average of the squared differences between the actual and predicted values, which penalizes larger errors more heavily and is useful when large mistakes are particularly undesirable. R-squared, also known as the coefficient of determination, indicates how well the regression model explains the variability of the dependent variable by showing the proportion of total variation in the outcome that is accounted for by the independent variable(s).


Q8. What is the purpose of the R-squared metric in regression analysis?

A8. The purpose of the R-squared metric in regression analysis is to measure how well a regression model explains the variation in the dependent variable. It indicates the proportion of the total variability in the outcome variable that is accounted for by the independent variable or variables included in the model. An R-squared value closer to 1 suggests that the model provides a good fit to the data, meaning it explains a large portion of the variation, while a value closer to 0 indicates that the model explains very little of the variation. R-squared is mainly used to assess the goodness of fit of a regression model and to compare how well different models explain the same dataset.


Q9.  Write Python code to fit a simple linear regression model using scikit-learn
and print the slope and intercept.
(Include your Python code and output in the code box below.)

A9.

```python
# Import necessary libraries
from sklearn.linear_model import LinearRegression
import numpy as np

# Example data
# X = independent variable (e.g., hours studied)
# y = dependent variable (e.g., exam scores)
X = np.array([2, 4, 6, 8, 10]).reshape(-1, 1)  # Reshape to 2D array for sklearn
y = np.array([50, 60, 65, 70, 80])

# Create a LinearRegression model
model = LinearRegression()

# Fit the model to the data
model.fit(X, y)

# Print the slope (coefficient) and intercept
print("Slope (β1):", model.coef_[0])
print("Intercept (β0):", model.intercept_)
```

Q10. How do you interpret the coefficients in a simple linear regression model?

A10. In a simple linear regression model, the coefficients represent the relationship between the independent variable and the dependent variable. There are two coefficients: the **intercept** ((β_0)) and the **slope** ((β_1)).

The **intercept ((β_0))** is the predicted value of the dependent variable ((Y)) when the independent variable ((X)) is zero. It represents the baseline or starting point of (Y) in the absence of (X). For example, if the intercept is 46, it means that when (X = 0), the expected value of (Y) is 46.

The **slope ((β_1))** indicates how much the dependent variable is expected to change for a **one-unit increase** in the independent variable. It shows both the direction and strength of the relationship. A positive slope means that as (X) increases, (Y) also increases, while a negative slope indicates that (Y) decreases as (X) increases. For example, if the slope is 3.64, it means that for every additional unit of (X), (Y) is expected to increase by 3.64 units on average.

In short, the intercept tells you **where the line starts**, and the slope tells you **how steep the line is** and the direction of the relationship between (X) and (Y).
